#Mission 7：建立專案模擬學生學習資料

```python
import numpy as np
import pandas as pd

n = 60
rng = np.random.default_rng(42)

student_ids = list(range(1, n + 1))
departments = rng.choice(["資訊管理系", "財務金融系", "人工智慧系"], size=n)
sleep_hours = np.clip(rng.normal(6.7, 1.1, n), 4.0, 10.0)
screen_time_hours = np.clip(rng.normal(5.0, 1.5, n), 1.0, 10.0)
study_hours = np.clip(rng.normal(2.5, 1.0, n), 0.5, 6.0)
attendance_rate = np.clip(rng.normal(88, 8, n), 60, 100)

df = pd.DataFrame({
    "student_id": student_ids,
    "department": departments,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

df.head()
```

#Mission 8：建立學期成績欄位

```python
raw_score = (
    50
    + (3 * df["study_hours"])
    + (2 * df["sleep_hours"])
    - (1.2 * df["screen_time_hours"])
    + (0.15 * df["attendance_rate"])
    + rng.normal(0, 5, len(df)) # 加入隨機誤差
)

# 將分數限制在 0 到 100 之間
df["quiz_score"] = np.clip(raw_score, 0, 100)

df[["student_id", "study_hours", "sleep_hours", "quiz_score"]].head()
```

#Mission 9：處理遺漏值與異常值

```python
# 人為製造遺漏值與異常值
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0

print("遺漏值與異常值植入完成！")
```

#Mission 10：執行一欄學習資料品質檢查

```python
print("--- 資料形狀 ---")
print(df.shape)

print("\n--- 遺漏值統計 ---")
print(df.isna().sum())

print("\n--- 統計描述 ---")
print(df.describe())
```

#Mission 11：篩選與查詢資料

```python
# 篩選成績大於或等於 80 分的學生
high_score_condition = df["quiz_score"] >= 80
high_score_students = df[high_score_condition]

print(f"成績 >= 80 分的學生人數：len(high_score_students)")
high_score_students.head()
```

```python
file_name = "student_learning_data.csv"

df.to_csv(
    file_name,
    index=False,
    encoding="utf-8-sig"
)

print("資料已成功匯出至：", file_name)

# 如果是在 Google Colab 環境，可使用以下程式碼下載：
try:
    from google.colab import files
    files.download(file_name)
except ImportError:
    print("非 Google Colab 環境，檔案已儲存在本機目錄。")
```
