```python
#W6-HW-1
import numpy as np
import pandas as pd

try:
    display
except NameError:
    try:
        from IPython.display import display
    except ImportError:
        display = print

student_count = 60
departments = np.array(["資訊工程", "企業管理", "電子工程", "機械工程", "應用外語"])
rng = np.random.default_rng(42)

# ===== 資料生成 =====
n = student_count
student_ids = np.arange(1, 61)
chosen_departments = rng.choice(departments, size=n)
sleep_values = rng.normal(6.7, 1.1, size=n).round(1)
screen_values = rng.normal(5.0, 1.5, size=n).clip(1, 10).round(1)
study_values = rng.normal(2.5, 1.0, size=n).clip(0.5, 6).round(1)
attendance_values = rng.normal(88, 8, size=n).clip(60, 100).round().astype(int)

df = pd.DataFrame({
    "student_id": student_ids,
    "department": chosen_departments,
    "sleep_hours": sleep_values,
    "screen_time_hours": screen_values,
    "study_hours": study_values,
    "attendance_rate": attendance_values,
})

# ===== 成績公式 =====
raw_scores = (50
              + 3 * df["study_hours"]
              + 2 * df["sleep_hours"]
              - 1.2 * df["screen_time_hours"]
              + 0.15 * df["attendance_rate"]
              + rng.normal(0, 5, size=n))

df["quiz_score"] = raw_scores.clip(0, 100).round().astype(int)

assert df["quiz_score"].between(0, 100).all(), "分數應在 0 到 100。"

# ===== 顯示全部 60 列 =====
with pd.option_context("display.max_rows", 60, "display.width", 200):
    display(df)
```
